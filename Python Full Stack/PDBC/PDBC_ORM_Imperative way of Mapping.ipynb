```python
from sqlalchemy import create_engine, MetaData, Table, Column, Integer, String, Float

from sqlalchemy.orm import registry , sessionmaker
```

```python
DB_URL =  'mysql+pymysql://root:pavaniswind@localhost:3306/sqlcls'

ENG = create_engine(DB_URL)
print(type(ENG))
```

```python
class Student:
    def __init__(self, roll, name, marks):
        self.roll = roll
        self.name = name
        self.marks = marks
        
    def __str__(self):
        return '\n\n Student Information'\
                f'\n Student roll: {self.roll}'\
                f'\n Student name: {self.name}'\
                f'\n Student marks: {self.marks}'\

md = MetaData()
tb = Table('stu_tb1', md,
          Column('roll', Integer, primary_key = True),
          Column('name', String(70)),
          Column('marks', Float))

mapper = registry()
mapper.map_imperatively(Student, tb)
md.create_all(ENG)
print('Table Create Sucessfully....')
```

```python
Session = sessionmaker(bind = ENG)
sess = Session()

sess.query(Student).delete() #If you're re-running your script multiple times during development/testing,
                             #it's handy to clear the table first:
sess.commit()

#Insert Data

s1 = Student(101, 'Ved', 90.21)

sess.add(s1) # add —> It is used to add single object in the table

s2 = Student(102,'Pavan', 92.45)
s3 = Student(103,'Vedang', 96.78)
s4 = Student(104,'Om', 98.56)

sess.add_all([s2,s3,s4]) #add_all —> To add Multiple objects

#Retrive Data

# result = sess.query(Student)
# for record in result:
#     print(record)

result = sess.query(Student).filter(Student.roll == 102)
for record in result:
    print(record)
sess.commit()
sess.close()
```
