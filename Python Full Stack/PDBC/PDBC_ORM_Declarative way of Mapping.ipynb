```python
from sqlalchemy import create_engine, Column, Integer, String, Float
from sqlalchemy.orm import declarative_base, sessionmaker
```

```python
DB_URL = 'mysql+pymysql://root:pavaniswind@localhost:3306/sqldecl'
ENG = create_engine(DB_URL)
Base = declarative_base()
```

```python
class Employee(Base):
    __tablename__ = 'emp_tb1'
    __table_args__ = {'extend_existing': True}
    
    eid = Column(Integer, primary_key = True)
    name = Column(String(70))
    sal = Column(Float)
    
    
    def __str__(self):
        return '\n Employee Information'\
              f'\n Employee ID: {self.eid}'\
              f'\n Employee name : {self.name}'\
              f'\n Employee salary: {self.sal}'

Base.metadata.create_all(ENG)
print('Table Created Sucessfully....')

Session = sessionmaker(bind = ENG)
sess = Session()


sess.query(Employee).delete() #If you're re-running your script multiple times during development/testing,
                             #it's handy to clear the table first:
sess.commit()

while 1 == 1:
    ch = int(input('\n MENU'\
                  '\n1. Insert Data'\
                  '\n2. Retrive Data'\
                  '\n3. Update Data'\
                  '\n4. Delete Data'\
                  '\n5. Exit'\
                  '\n Enter your Choice'))
    
    #INSERT DATA
    if ch == 1:
        n = int(input("\n How many Records you want to ADD: "))
        employees_list = []
        
        for i in range(n):
            eid = int(input('\n Enter Employee ID: '))
            name = input('\n Enter Employee name: ')
            sal = float(input('\n Enter Employee salary: '))
            
            obj = Employee(eid = eid, name = name, sal = sal)
            employees_list.append(obj)
            
        sess.add_all(employees_list)
        
    #RETRIVE DATA
    elif ch == 2:
        ch1 = int(input('\n You want to Retrive'\
                       '\n1. All Records'\
                       '\n2. Particular Record'\
                       '\n Enter your Choice'))
        
        # RETRIVE ALL RECORDS
        if ch1 == 1:
            result = sess.query(Employee)
            for record in result:
                print(record)
                
        # RETRIVE PARTICULAR RECORD
        elif ch1 == 2:
            eid_ref = int(input('\n Enter Employee ID whose record you want to Retrive'))
            result = sess.query(Employee).filter(Employee.eid == eid_ref)
            
            for record in result:
                print(record)
                
    #UPDATE DATA
    elif ch == 3:
        eid_ref = int(input('\n Enter Employee ID whose record you want to update:'))
        ch1 = int(input('\n What do you want to Update:'\
                       '\n1. Employee Name: '\
                       '\n2. Employee Salary: '\
                       '\n Enter your choice: '))
        
        # TO UPDATE NAME
        if ch1 == 1:
            name_new = input('\n Enter New Name To Update: ')
            sess.query(Employee).filter(Employee.eid == eid_ref).update({Employee.name : name_new})
            
        # TO UPDATE SALARY
        elif ch1 == 2:
            sal_new = float(input('\n Enter New Salary To Update: '))
            sess.query(Employee).filter(Employee.eid == eid_ref).update({Employee.sal : sal_new})
            
    # DELETE DATA
    elif ch == 4:
        ch1 = int(input('\n What do you want to Delete:'\
                       '\n1. All Records'\
                       '\n2. Particular Record'\
                       '\n Enter your choice'))
        
        #DELETE ALL RECORD
        if ch1 == 1:
            sess.query(Employee).delete()
            
        # DELETE PARTICULAR RECORD
        if ch1 == 2:
            eid_ref = int(input('\n Enter Employee ID whose record you want to Delete: '))
            
            sess.query(Employee).filter(Employee.eid == eid_ref).delete()
            
    # EXIT
    elif ch == 5:
        break;
    
    # DEFAULT BLOCK
    else:
        print('\n\n\t INVALID OPTION...!!')
        
    sess.commit()
sess.close()
```
